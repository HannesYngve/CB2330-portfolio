The formula calculated the fraction of G or C in a sequence.
1/L ∑ L,i=1 = 1[bi∈{G,C}]

L is the length of sequence. To get a fraction it is 1/L

∑ L,i=1 is the sum of all from 1(i) to L.

bi∈{G,C} is that bi is an element of the set which consists of G an C.

In a sentence it takes the length of the sequence and counts the C or G and calculates divides the found CG with the sequence length.

```python
import numpy as np

#First with base python loop.
seq = 'ACGTAGCTAGTCGCGTCGTCGAGCA'
foundGC = 0
for char in seq:
  if char == 'G' or char == 'C':
    foundGC =foundGC + 1
GCs = foundGC/len(seq)
print("found base GC: ",foundGC, "Total nucleotide: ",len(seq), "GC%: ",GCs )

#Now using numpy

gc_content = (np.char.count(seq, sub = 'G') + np.char.count(seq, sub = 'C')) / len(seq)
print("numpy GC: ", gc_content)
```
